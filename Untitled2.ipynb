```python
# ==========================================
# PROJECT: DATA CLEANING AND VISUALIZATION
# ==========================================

# STEP 1: IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")


# ==========================================
# STEP 2: CREATE DATASET
# ==========================================

data = {
    'Name': ['Ravi', 'Sita', 'Ram', 'John',
             'Priya', 'Kiran', 'Anu', 'Ravi'],
    'Marks': [85, 92, np.nan, 75, 88, 65, 78, 85],
    'Attendance': [90, 95, 80, 85, 92, 70, 88, 90],
    'Study_Hours': [5, 6, 4, 3, 5, 2, 4, 5]
}

df = pd.DataFrame(data)

print("\n ORIGINAL DATASET")
print(df)


# ==========================================
# STEP 3: UNDERSTAND THE DATA
# ==========================================

print("\n FIRST FIVE ROWS")
print(df.head())

print("\n DATA INFORMATION")
df.info()

print("\n STATISTICAL DESCRIPTION")
print(df.describe())


# ==========================================
# STEP 4: CHECK MISSING VALUES
# ==========================================

print("\n MISSING VALUES")
print(df.isnull().sum())

# Fill missing marks with median
df['Marks'] = df['Marks'].fillna(
    df['Marks'].median()
)

print("\n AFTER HANDLING MISSING VALUES")
print(df)


# ==========================================
# STEP 5: REMOVE DUPLICATE VALUES
# ==========================================

print("\n DUPLICATE ROWS:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\n AFTER REMOVING DUPLICATES")
print(df)


# ==========================================
# STEP 6: HANDLE OUTLIERS USING IQR
# ==========================================

Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("\n LOWER LIMIT:", lower_limit)
print("UPPER LIMIT:", upper_limit)

# Identify outliers
outliers = df[(df['Marks'] < lower_limit) | (df['Marks'] > upper_limit)]
print("\nOUTLIERS FOUND:")
print(outliers)
```
